```python
# function for reading string list txt file into list of list with int instead of strings: 
def read_into_list_int(file_path):
    point = []
    with open(file_path, 'r') as infile:
        for line in infile:
            line = line.rstrip().split(",")
            x = []
            for num in line:
                x.append(int(num))
            point.append(x)
    return point
        
```
