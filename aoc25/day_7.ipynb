```python
import numpy as np 
```

```python
# def function to read txt file into lines 
def read_file_into_lines(file_path):
    lines = []
    with open(file_path, 'r') as infile:
        for line in infile.readlines():
            cleaned = line.rstrip()
            lines.append(cleaned)
    return lines
```

```python
# use function for data set
raw = read_file_into_lines("C:/Users/gabia/Desktop/python_homework_gabi/data/advent_of_code_day_7.txt")
```

```python
manifold = np.array([list(x)for x in raw])
```

```python
# function to get position of beam starting
def start_of_beam(manifold):
    for i, x in enumerate(manifold[0]):
        if x == 'S':
            return i
    return None
```

```python
def beam_line_count(incoming, line):
    # incoming 
    outgoing = []
    split_count = 0 
    beam = line.copy()
    
    # can we split?
    for spot in incoming:
        if line[spot] == '.':
            if spot not in outgoing: 
                outgoing.append(spot)
            beam[spot] = '|'
       
        else:
            split_count += 1
            
             # make new lines
            left = spot - 1
            right = spot + 1
            
            # go over left & right
            if left > -1 and left not in outgoing:
                beam[left] = '|'
                outgoing.append(left)
                
            if right < len(line) and right not in outgoing:
                beam[right] = '|'
                outgoing.append(right)
    # print(split_count)
    return split_count, outgoing
```

```python
def all_splits(manifold):
    start = start_of_beam(manifold)
    incoming = [start]
    all_splits = 0
    for line in manifold[1:]:
        split_count, outgoing = beam_line_count(incoming, line)
        all_splits += split_count
        incoming = outgoing

    return all_splits
```

```python
all_splits(manifold)
```
